```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Month_Name'] = df['Order Date'].dt.month_name()
```

```python
monthly_sales = df.groupby(
    ['Year', 'Month', 'Month_Name']
)['Sales'].sum().reset_index()
monthly_sales.head(15)
```

```python
monthly_sales = monthly_sales.sort_values(
    by=['Year', 'Month']
)
monthly_sales
```

```python
plt.figure(figsize=(14, 6))
plt.plot(
    monthly_sales['Month_Name'] + ' ' + monthly_sales['Year'].astype(str),
    monthly_sales['Sales'],
    marker='o'
)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
highest_month=monthly_sales.loc[monthly_sales['Sales'].idxmax()]
lowest_month=monthly_sales.loc[monthly_sales['Sales'].idxmin()]
print("Highest Sales Month:")
print(highest_month)
print("\nLowest Sales Month:")
print(lowest_month)
```

```python
customer_sales=df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False)
print("Top 10 Customers by Sales:")
customer_sales.head(10)
```

```python
customer_profit=df.groupby('Customer Name')['Profit'].sum().sort_values(ascending=False)
print("Top 10 Customers by Profit:")
customer_profit.head(10)
```

```python
top_customers = customer_sales.head(10)
plt.figure(figsize=(12, 6))
top_customers.sort_values().plot(kind='barh')
plt.title('Top 10 Customers by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Customer Name')
plt.tight_layout()
plt.show()
```

```python
discount_analysis=df.groupby('Discount').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Quantity=('Quantity', 'sum'),
    Total_Transactions=('Order ID', 'count')
).reset_index()
discount_analysis
```

```python
discount_analysis['Profit_Margin'] = (
    discount_analysis['Total_Profit'] /
    discount_analysis['Total_Sales']
) * 100
discount_analysis
```

```python
plt.figure(figsize=(10, 6))
plt.bar(
    discount_analysis['Discount'].astype(str),
    discount_analysis['Total_Profit']
)
plt.title('Impact of Discount on Total Profit')
plt.xlabel('Discount Level')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(
    discount_analysis['Discount'],
    discount_analysis['Profit_Margin'],
    marker='o'
)
plt.title('Discount vs Profit Margin')
plt.xlabel('Discount Level')
plt.ylabel('Profit Margin (%)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

Recommendation:Review discount policies and optimize promotional pricing to maintain business profitability.
